````python
!pip install python-dotenv
!pip install openai
!pip install gradio
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
````

````python
# Load environment variables
load_dotenv()

openai_api_key = "REDACTED_OPENAI_API_KEY"

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    raise ValueError("OPENAI_API_KEY not set in .env file")
````

````python
MODEL = "gpt-4o-mini"
client = OpenAI(api_key=openai_api_key)

system_message = (
    "You are the File Handling Assistant in Knowhere cloud storage service. "
    "You can create files and images. "
    "If the user asks about current events, news, or information not in your knowledge, "
    "use the web_search tool."
)
````

````python
def create_text_file(filename,content):
    with open(filename, "w", encoding="utf-8") as file:
        file.write(content)
    return f"Text file '{filename}' created successfully."
````

````python

def create_pdf(filename, content):
    messages = [
        {
            "role": "system",
            "content": (
                "Generate only valid Python code that creates a colorful PDF file. "
                "Return code only. No explanation. No markdown fences."
            )
        },
        {
            "role": "user",
            "content": f"Create Python code to generate a PDF file named '{filename}' with this content: {content}"
        }
    ]

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
    )

    generated_code = response.choices[0].message.content.strip()
    generated_code = generated_code.replace("```python", "").replace("```", "").strip()

    print("Generated code:\n", generated_code)

    exec(generated_code, {"filename": filename, "content": content})
    return f"PDF file '{filename}' created successfully."
````

````python
from duckduckgo_search import DDGS

def web_search(query):
    try:
        results = []
        with DDGS() as ddgs:
            search_results = ddgs.text(query, max_results=5)
            for r in search_results:
                results.append({
                    "title": r.get("title", ""),
                    "link": r.get("href", ""),
                    "snippet": r.get("body", "")
                })
        return results
    except Exception as e:
        print("WEB SEARCH ERROR:", repr(e))
        return [{"error": str(e)}]
````

````python

import requests
import time

def create_image(filename, content, size="1024x1024"):
    valid_sizes = ["1024x1024", "1024x1792", "1792x1024"]
    if size not in valid_sizes:
        size = "1024x1024"  # ✅ fix before hitting API

    max_retries = 3
    for attempt in range(max_retries):
        try:
            image_response = client.images.generate(
                model="dall-e-3",
                prompt=content,
                size=size,
                n=1
            )
            image_url = image_response.data[0].url
            img_data = requests.get(image_url).content
            with open(filename, "wb") as f:
                f.write(img_data)
            return f"Image '{filename}' created successfully."
        except Exception as e:
            print(f"Attempt {attempt+1} failed: {e}")
            if attempt < max_retries - 1:
                time.sleep(2)
    return "Image creation failed after 3 attempts."
 
````

````python
filecreate_function = {
    "name": "create_text_file",
    "description": "Create a text file with the content requested by the user.",
    "parameters": {
        "type": "object",
        "properties": {
                "filename": {
                "type": "string",
                "description": "provide the file with relavant extention(.html,.txt etc.) but do not added pdf format files for this"
            },
            "content": {
                "type": "string",
                "description": "The content to write into the text file"
            }
        },
        "required": ["filename","content"],
        "additionalProperties": False
    }
}

websearch_function = {
    "name": "web_search",
    "description": "Search the web for current information using DuckDuckGo",
    "parameters": {
        "type": "object",
        "properties": {
            "query": {
                "type": "string",
                "description": "Search query"
            }
        },
        "required": ["query"]
    }
}
pdfcreate_function = {
    "name": "create_pdf",
    "description": "Create a pdf file with the content requested by the user.",
    "parameters": {
        "type": "object",
        "properties": {
                "filename": {
                "type": "string",
                "description": "provide the filename with .pdf  but do not added pdf format files for this"
            },
            "content": {
                "type": "string",
                "description": "The content to write into the pdf file"
            }
        },
        "required": ["filename","content"],
        "additionalProperties": False
    }
}

imagecreate_function = {
    "name": "create_image",
    "description": "Create an image with the content requested by the user.",
    "parameters": {
        "type": "object",
        "properties": {
            "filename": {
                "type": "string",
                "description": "Filename with relevant image extension e.g. image.png"
            },
            "content": {
                "type": "string",
                "description": "The prompt to generate the image"
            },
            "size": {
                "type": "string",
                "enum": ["1024x1024", "1024x1792", "1792x1024"],  # ✅ restrict choices
                "description": "Image size. Must be one of: 1024x1024, 1024x1792, 1792x1024"
            }
        },
        "required": ["filename", "content", "size"],
        "additionalProperties": False
    }
}
````

````python
tools = [
    {"type": "function", "function": filecreate_function},
    {"type": "function", "function": pdfcreate_function},
    {"type": "function", "function": imagecreate_function},
    {"type": "function", "function": websearch_function}
         ]
````

````python
def handle_tool_call(message):
    tool_call = message.tool_calls[0]
    arguments = json.loads(tool_call.function.arguments)
    tool_name = tool_call.function.name

    if tool_name == "create_text_file":
        result = create_text_file(arguments["filename"], arguments["content"])

    elif tool_name == "create_pdf":
        result = create_pdf(arguments["filename"], arguments["content"])

    elif tool_name == "create_image":
        size = arguments.get("size", "1024x1024")
        result = create_image(arguments["filename"], arguments["content"], size)

    elif tool_name == "web_search":
        result = web_search(arguments["query"])

    else:
        result = f"Unknown tool: {tool_name}"

    return {
        "role": "tool",
        "tool_call_id": tool_call.id,
        "content": json.dumps({"result": result}, ensure_ascii=False)
    }
````

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_message}]

    # ✅ Handle both old tuple format and new Gradio dict format
    for item in history:
        if isinstance(item, dict):
            messages.append({"role": item["role"], "content": item["content"]})
        else:
            user_msg, assistant_msg = item
            if user_msg:
                messages.append({"role": "user", "content": user_msg})
            if assistant_msg:
                messages.append({"role": "assistant", "content": assistant_msg})

    messages.append({"role": "user", "content": message})

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        tools=tools
    )
    assistant_message = response.choices[0].message

    if response.choices[0].finish_reason == "tool_calls":
        tool_response = handle_tool_call(assistant_message)
        messages.append(assistant_message)
        messages.append(tool_response)

        final_response = client.chat.completions.create(
            model=MODEL,
            messages=messages
        )
        return final_response.choices[0].message.content

    return assistant_message.content
````

````python

# Gradio UI
demo = gr.ChatInterface(fn=chat, title="Text File Creator Assistant")

demo.launch()
````
